```python
#=================1.Text to tokens=================


import nltk

# NLTK-ku thevaiyana punctuation matrum word split rules-a download pandrathuku:
nltk.download('punkt') 

from nltk.tokenize import word_tokenize
import nltk

my_text = "Machine learning is awesome! But, isn't it hard?"
tokens = word_tokenize(my_text)

print(tokens)


#=================2.Stop words=================

nltk.download('stopwords')
from nltk.corpus import stopwords

# English stop words list-a edukkuroam
stop_words = set(stopwords.words('english'))

filtered_words = []
for word in tokens:
    # Stop words matrum punctuations (isalpha use panni) thedi remove pandroam
    if word.lower() not in stop_words and word.isalpha():
        filtered_words.append(word)

print(filtered_words)

```

```python
import nltk
# Lemmatization-ku thevaiyana dictionary files download pandrom
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Rendu tools-aiyum ready pandrom
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Namma test panna pora 3 words
words = ['learning', 'studies', 'universities']

print("--- STEMMING (Vetti edukkum process) ---")
for word in words:
    print(f"{word} -> {stemmer.stem(word)}")

print("\n--- LEMMATIZATION (Dictionary process) ---")
for word in words:
    print(f"{word} -> {lemmatizer.lemmatize(word)}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Namma kitta irukka 2 chinna reviews (Clean panna text)
reviews = [
    "awesome movie",
    "worst battery"
]

# Vectorizer tool-a ready pandrom (Idhudhaan namma Checklist uruvakkum tool)
vectorizer = CountVectorizer()

# Text-a ulla anuppi, dictionary ready panni, numbers-a maathurom
vectors = vectorizer.fit_transform(reviews).toarray()

# AI uruvakkina Master Checklist (Dictionary words)
print("Master Checklist Words:", vectorizer.get_feature_names_out())

# Vector aana numbers (0s and 1s):
print("\nVectors (Numbers):")
print(vectors)
```
