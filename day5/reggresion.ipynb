```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Namba Data: Veedoda Size (sq.ft) vs Price (Lakhs-la)
data = {
    'Size': [600, 800, 1000, 1200, 1500],
    'Price': [35, 45, 55, 65, 80]
}
df = pd.DataFrame(data)

# 2. X (Clues) matrum y (Answer) pirikkurom
X = df[['Size']] 
y = df['Price']  

# 3. Model-a uruvakki Train pandrom
model = LinearRegression()
model.fit(X, y)

# 4. Pudhu veetuku price predict pandrom!
pudhu_veedu = [[1300]]
predicted_price = model.predict(pudhu_veedu)

print(f"1300 sq.ft veedoda price (predicted): {predicted_price[0]:.2f} Lakhs")
import matplotlib.pyplot as plt

# 1. Unmaiyana Data points-a (veedu size vs price) scatter plot-la (pulligala) vekkurom
plt.scatter(X, y, color='blue', label='Unmaiyana Data')

# 2. Namba AI model predict panna 'Line of Best Fit'-a varaiyurom
predicted_y = model.predict(X)
plt.plot(X, predicted_y, color='red', label='AI Predict panna Line')

plt.xlabel('Veetoda Size (sq.ft)')
plt.ylabel('Price (Lakhs)')
plt.title('Linear Regression: Veedu Size vs Price')
plt.legend()
plt.show()
```
