```python
import pandas as pd
from datasets import load_dataset

print("Hugging Face-la irundhu data download aagudhu... Please wait ⏳")

# 1. Hugging Face server-la irundhu dataset-a load pandrom.
dataset = load_dataset("sms_spam")

# 2. Hugging Face format-a namma pazhaiya Pandas DataFrame format-ku maathurom
df = dataset['train'].to_pandas()

# 3. Columns-ku correct-aana peru vechu, clean pandrom
df = df[['label', 'sms']]
df.columns = ['label', 'message'] 

# Label numbers-a text-a maathidalam (0 -> ham, 1 -> spam) namma code logic-kaga
df['label'] = df['label'].map({0: 'ham', 1: 'spam'})

print("\nData Load Aagiduchu! 🎉")
print(f"Total Rows: {df.shape[0]}\n")
print(df.head())

# 4. DataFrame-a CSV file aaga save pandrom
df.to_csv('spam.csv', index=False)
print("\nSuccess! 'spam.csv' file unga folder-la save aagiduchu.")
```
